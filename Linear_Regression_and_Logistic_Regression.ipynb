```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
hours = np.array([2, 4, 6, 8, 10]).reshape(-1,1)
marks = np.array([55, 60, 70, 80, 88])

display(hours, marks)
```

```python
model = LinearRegression()
model.fit(hours, marks)
```

```python
print(model.coef_[0])
print(model.intercept_)
```

```python
# y = 4.3x + 44.8
plt.scatter(hours, marks, color='blue', label='Actual Data')

predicted_marks = model.predict(hours)
plt.plot(hours, predicted_marks, color='lime', label='Predicted Line')

plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Study Hours vs Marks for linear regression')
plt.legend()
plt.grid(True)
plt.show()
```

```python
new_hour = np.array([[7]])
prediction = model.predict(new_hour)
print(f"Predicted Marks: ", prediction[0])
```

```python
from sklearn.linear_model import SGDRegressor
from sklearn.metrics import mean_squared_error

# compute mean square error
mse = mean_squared_error(marks, predicted_marks)
print("MSE for linear regresssion: ", mse)
```

# Stochastic Gradiend Descent(SGDRegressor)

```python
sgd_reg = SGDRegressor(
    max_iter=3000,
    learning_rate='invscaling',
    eta0=0.01,
    random_state=42
)

sgd_reg.fit(hours, marks)
print('m', sgd_reg.coef_)
print('c', sgd_reg.intercept_)

predicted_sgd_marks = sgd_reg.predict(hours)
mse_sgd = mean_squared_error(marks, predicted_sgd_marks)
print("MSE for SGDRegressor: ", mse_sgd)
```

```python
# y = 4.3x + 44.8
plt.scatter(hours, marks, color='blue', label='Actual Data')

plt.plot(hours, predicted_marks, linestyle='--')
plt.plot(hours, predicted_sgd_marks, linestyle='-.')

plt.xlabel('Study Hours')
plt.ylabel('Marks')
plt.title('Study Hours vs Marks for linear regression')
plt.legend()
plt.grid(True)
plt.show()
```
